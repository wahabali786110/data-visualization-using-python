```python
import seaborn as snc
import matplotlib.pyplot as plt
import plotly.express as px
```

# 1. Relational Plots

Scatter Plot

```python
tips=snc.load_dataset('tips')
tips.head()
```

```python
# Using axes level function
snc.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
# Using figure level function
snc.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

Line plot

```python
gap_data=px.data.gapminder()
temp_df=gap_data[gap_data['country']=='Pakistan']
temp_df.head()
```

```python
snc.lineplot(data=temp_df,x='year',y='lifeExp')
```

```python
new_temp_df=gap_data[gap_data['country'].isin(['Pakistan','Brazil','Germany'])]
```

```python
snc.relplot(kind='line',data=new_temp_df,x='year',y='lifeExp',hue='country',style='continent')
```

Facet Plot

```python
snc.relplot(data=tips,x='total_bill',y='tip',kind='scatter',col='smoker',row='sex')
```

```python
gap_data
```

```python
# col_wrap
snc.relplot(kind='scatter',data=gap_data,x='lifeExp',y='gdpPercap',col='year',col_wrap=3)
```

# 2. Distribution Plots

Hist Plot

```python
snc.histplot(data=tips,x='total_bill',hue='sex',element='step',bins=20)
```

```python
titanic=snc.load_dataset('titanic')

snc.displot(kind='hist',data=titanic,x='age',col='sex',element='step')
```

KDE plot

```python
snc.kdeplot(data=tips,x='total_bill',hue='sex',fill=True)
```

```python
snc.displot(kind='kde',data=tips,x='total_bill',fill=True)
```

Rug plot

```python
snc.kdeplot(data=tips,x='total_bill')
snc.rugplot(data=tips,x='total_bill')
```

Bivariate Histplot and Kde plot

```python
# Hist Plot
snc.displot(kind='hist',data=tips,x='total_bill',y='tip')
```

```python
# Kde Plot
snc.displot(kind='kde',data=tips,x='total_bill',y='tip')
```

# 3. Matrix Plots

Heatmap

```python
temp_df=gap_data[gap_data['continent']=='Asia'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(15,15))
snc.heatmap(temp_df,linewidths='0.5',annot=True,cmap='viridis')
```

Clustermap

```python
iris=px.data.iris()
iris.head()
```

```python
snc.clustermap(iris.iloc[:,[0,1,2,3]])
```
